**24. Merging and concatenating (2 pts)**

Given:
```python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"],
                         "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"],
                         "grade": [90, 85, 88, 92]})
```
- (a) Merge `students` and `grades` using an **inner** join. Print the result.
- (b) Merge `students` and `grades` using a **left** join. Print the result and add a comment explaining why Ben has `NaN`.
- (c) Use `pd.concat()` to add a new row for `"Ella"` in section `"A"` to `students`. Print the result.

````python
students = pd.DataFrame({"name": ["Ana", "Ben", "Carla", "Dan"],
                         "section": ["A", "B", "A", "B"]})
grades   = pd.DataFrame({"name": ["Ana", "Carla", "Dan", "Ella"],
                         "grade": [90, 85, 88, 92]})

merged_inner = pd.merge(students, grades, on='name', how='inner')
print("Merged (inner) DataFrame:")
print(merged_inner)

merged_left = pd.merge(students, grades, on='name', how='left')
print("\nMerged (left) DataFrame:")
print(merged_left)

new_row = pd.DataFrame({"name": ["Ella"], "section": ["A"]})
students_with_ella = pd.concat([students, new_row], ignore_index=True)
print("\nStudents with Ella:")
print(students_with_ella)
````

**26. Connect and create a table (5 pts)**

- (a) Import `sqlite3` and connect to a database named `'school.db'`. Save the connection in a variable named `conn` and print `"Opened database successfully"`.
- (b) Run this line first so the cell can be re-run without errors (it deletes the old table if it already exists):
  ```python
  conn.execute('DROP TABLE IF EXISTS STUDENTS')
  ```
- (c) Create a table named `STUDENTS` with these columns, then print `"Table created successfully"`:

| Column | Type | Rule |
|---|---|---|
| `ID` | `INT` | `PRIMARY KEY`, `NOT NULL` |
| `NAME` | `TEXT` | `NOT NULL` |
| `AGE` | `INT` | `NOT NULL` |
| `COURSE` | `CHAR(50)` | |
| `GWA` | `REAL` | |

````python
import sqlite3 as sql

conn = sql.connect('school.db')
print("Opened database successfully")
conn.execute('DROP TABLE IF EXISTS STUDENTS')
conn.execute('''
    CREATE TABLE STUDENTS (
        ID INT PRIMARY KEY NOT NULL,
        NAME TEXT NOT NULL,
        AGE INT NOT NULL,
        COURSE CHAR(50),
        GWA REAL
    )
''')
print("Table created successfully")
````

| ID | NAME | AGE | COURSE | GWA |

**28. Update and delete records (5 pts)**

- (a) **Update** Carla's record (`ID = 3`) so that her `GWA` becomes `1.25`. Commit the change.
- (b) **Delete** Dan's record (`ID = 4`). Commit the change.
- (c) Print `conn.total_changes`.
- (d) Select `ID`, `NAME` and `GWA` from `STUDENTS` and print every remaining record using a `for` loop.
- (e) Close the connection.

````python
import sqlite3
conn = sqlite3.connect('school.db')

conn.execute("UPDATE STUDENTS SET GWA = 1.25 WHERE ID = 3")
conn.commit()
conn.execute("DELETE FROM STUDENTS WHERE ID = 4")
conn.commit()
print(f"Total changes: {conn.total_changes}")

cursor = conn.execute("SELECT ID, NAME, GWA FROM STUDENTS")
for row in cursor:
    print(f"ID = {row[0]}, NAME = {row[1]}, GWA = {row[2]}")

conn.close()
````
